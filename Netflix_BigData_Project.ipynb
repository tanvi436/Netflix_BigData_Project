```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("NetflixBigDataAnalysis") \
    .getOrCreate()

print("Spark Started Successfully!")
```

```python
df = spark.read.csv("dataset/netflix_titles.csv", header=True, inferSchema=True)

df.show(5)
df.printSchema()
```

```python
df.groupBy("type").count().show()
```

```python
df.groupBy("country").count().orderBy("count", ascending=False).show(10)
```

```python
df.groupBy("rating").count().orderBy("count", ascending=False).show(10)
```

```python
df_cleaned = df.fillna({
    "director": "Unknown",
    "cast": "Unknown",
    "country": "Unknown",
    "rating": "Not Rated"
})

df_cleaned.show(5)
```

```python
df_cleaned.write.mode("overwrite").option("header", "true").csv("output/netflix_cleaned_csv")

print("Cleaned CSV saved successfully!")
```

```python
import os
import glob
import shutil

final_path = "output/netflix_cleaned_final.csv"

# Delete if exists
if os.path.exists(final_path):
    os.remove(final_path)

# Get largest part file
csv_parts = glob.glob("output/netflix_cleaned_csv/part-*.csv")
largest_file = max(csv_parts, key=os.path.getsize)

shutil.copy(largest_file, final_path)

print("Final CSV created:", final_path)
```
